```python
import evaluate
import torch
import re
from transformers import MarianMTModel, MarianTokenizer
from unidecode import unidecode

# ✅ Load evaluation metrics
bleu = evaluate.load("bleu")
rouge = evaluate.load("rouge")
meteor = evaluate.load("meteor")
comet = evaluate.load("comet")
```

```python

adapter_path_1 = "/content/drive/MyDrive/Spanish_English/Models/fine_tuned_model_1/checkpoint-4000"
adapter_path_2 = "/content/drive/MyDrive/Spanish_English/Models/fine_tuned_model_2"
adapter_path_3 = "/content/drive/MyDrive/Spanish_English/Models/fine_tuned_model_3"
```

```python
original_model = MarianMTModel.from_pretrained(model_name)

fine_tuned_model_1 = MarianMTModel.from_pretrained(adapter_path_1)
fine_tuned_model_2 = MarianMTModel.from_pretrained(adapter_path_2)
fine_tuned_model_3 = MarianMTModel.from_pretrained(adapter_path_3)
```

```python

def translate(model, sentences):
    inputs = tokenizer(sentences, return_tensors="pt", padding=True, truncation=True)
    with torch.no_grad():
        outputs = model.generate(**inputs)
    return [tokenizer.decode(output, skip_special_tokens=True) for output in outputs]
```

```python

def compute_bleu(references, predictions):
    return bleu.compute(predictions=predictions, references=[[ref] for ref in references])["bleu"]

def compute_meteor(references, predictions):
    return meteor.compute(predictions=predictions, references=references)["meteor"]

def compute_rouge(references, predictions):
    rouge_scores = rouge.compute(predictions=predictions, references=references)
    return rouge_scores["rougeL"]

def compute_comet(sources, predictions, references):
    results = comet.compute(
        sources=sources, predictions=predictions, references=references
    )
    return results["mean_score"]
```

```python
import pandas as pd

def compute_weighted_average(scores, weights):
    """Computes the weighted average of scores based on test sentence counts."""
    weighted_sum = sum(score * weight for score, weight in zip(scores, weights))
    return weighted_sum / sum(weights)

test_set_sizes = [len(test_sentences), len(test_sentences_1)]

models = {
    "Original Model": [original_translations, original_translations_1, original_translations_2, original_translations_3],
    "Fine-Tuned Model 1": [fine_tuned_1_translations, fine_tuned_1_translations_1,  fine_tuned_1_translations_2,  fine_tuned_1_translations_3],
    "Fine-Tuned Model 2": [fine_tuned_2_translations, fine_tuned_2_translations_1, fine_tuned_2_translations_2, fine_tuned_2_translations_3],
    "Fine-Tuned Model 3": [fine_tuned_3_translations, fine_tuned_3_translations_1,  fine_tuned_3_translations_2,  fine_tuned_3_translations_3]
}

test_sets = [
    {"sentences": test_sentences, "references": reference_translations},
    {"sentences": test_sentences_1, "references": reference_translations_1},
    {"sentences": test_sentences_2, "references": reference_translations_2},
    {"sentences": test_sentences_3, "references": reference_translations_3}
]

results = []

for model_name, translations_list in models.items():
    bleu_scores, meteor_scores, rouge_scores, comet_scores = [], [], [], []

    for i, test_set in enumerate(test_sets):
        references = test_set["references"]
        translations = translations_list[i]

        bleu_scores.append(compute_bleu(references, translations))
        meteor_scores.append(compute_meteor(references, translations))
        rouge_scores.append(compute_rouge(references, translations))
        comet_scores.append(compute_comet(test_set["sentences"], translations, references))

    bleu_avg = compute_weighted_average(bleu_scores, test_set_sizes)
    meteor_avg = compute_weighted_average(meteor_scores, test_set_sizes)
    rouge_avg = compute_weighted_average(rouge_scores, test_set_sizes)
    comet_avg = compute_weighted_average(comet_scores, test_set_sizes)

    results.append([model_name, bleu_avg, meteor_avg, rouge_avg, comet_avg])

df = pd.DataFrame(results, columns=["Model", "Weighted BLEU", "Weighted METEOR", "Weighted ROUGE", "Weighted COMET"])

print("\n=== TRANSLATION QUALITY COMPARISON TABLE ===\n")
print(df.to_string(index=False))
```
